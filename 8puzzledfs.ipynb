```python
# 8-Puzzle using DFS

start = [
    [1, 2, 3],
    [0, 4, 6],
    [7, 5, 8]
]

goal = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 0]
]


# Print the puzzle
def print_puzzle(state):
    for row in state:
        print(row)
    print()


# Convert list to tuple
# Used for storing states in visited
def make_tuple(state):
    return tuple(tuple(row) for row in state)


# Find the position of 0
def find_zero(state):
    for i in range(3):
        for j in range(3):
            if state[i][j] == 0:
                return i, j


# Move 0 to a new position
def move(state, new_i, new_j):

    i, j = find_zero(state)

    # Make a copy
    new_state = [row[:] for row in state]

    # Swap 0 with the selected number
    new_state[i][j], new_state[new_i][new_j] = \
        new_state[new_i][new_j], new_state[i][j]

    return new_state


# Generate all possible moves
def get_neighbors(state):

    i, j = find_zero(state)

    neighbors = []

    # Move UP
    if i > 0:
        neighbors.append(move(state, i - 1, j))

    # Move DOWN
    if i < 2:
        neighbors.append(move(state, i + 1, j))

    # Move LEFT
    if j > 0:
        neighbors.append(move(state, i, j - 1))

    # Move RIGHT
    if j < 2:
        neighbors.append(move(state, i, j + 1))

    return neighbors


# DFS
def dfs(start, goal):

    # Stack contains (state, depth)
    stack = [(start, 0)]

    # Store visited states
    visited = set()

    while stack:

        # Take the last element from stack
        state, depth = stack.pop()

        # Convert state to tuple
        state_tuple = make_tuple(state)

        # Skip if already visited
        if state_tuple in visited:
            continue

        # Mark as visited
        visited.add(state_tuple)

        print("Depth:", depth)
        print_puzzle(state)

        # Check goal
        if state == goal:
            print("GOAL FOUND!")
            print("Depth:", depth)
            return

        # Generate next states
        neighbors = get_neighbors(state)

        # Add children to stack
        for next_state in neighbors:

            next_tuple = make_tuple(next_state)

            if next_tuple not in visited:
                stack.append((next_state, depth + 1))

    print("Goal not found")


# Start DFS
dfs(start, goal)
```
