```python
# Vacuum Cleaner Agent

room_A = "dirty"
room_B = "dirty"

# User gives only the starting position
position = input("Enter vacuum position (A/B): ").upper()

# Agent keeps working until both rooms are clean
while True:

    print("\n----------------------")
    print("Room A:", room_A)
    print("Room B:", room_B)
    print("Vacuum Position:", position)

    # AGENT PERCEIVES THE CURRENT ROOM
    if position == "A":

        if room_A == "dirty":
            # Agent decides to SUCK
            print("Agent: Room A is dirty.")
            print("Agent action: SUCK")
            room_A = "clean"

        else:
            # Room A is clean, so agent moves RIGHT
            print("Agent: Room A is clean.")
            print("Agent action: RIGHT")
            position = "B"

    elif position == "B":

        if room_B == "dirty":
            # Agent decides to SUCK
            print("Agent: Room B is dirty.")
            print("Agent action: SUCK")
            room_B = "clean"

        else:
            # Room B is clean, so agent moves LEFT
            print("Agent: Room B is clean.")
            print("Agent action: LEFT")
            position = "A"

    # GOAL TEST
    if room_A == "clean" and room_B == "clean":
        print("\n----------------------")
        print("Both rooms are clean!")
        print("Goal state reached.")
        break
```
