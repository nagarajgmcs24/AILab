```python
# 8-Puzzle using Iterative Deepening Search (IDS)

start = [
    [1, 2, 3],
    [0, 4, 6],
    [7, 5, 8]
]

goal = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 0]
]


# Print the puzzle
def print_puzzle(state):
    for row in state:
        print(row)
    print()


# Convert state into tuple
def make_tuple(state):
    return tuple(tuple(row) for row in state)


# Find position of 0
def find_zero(state):
    for i in range(3):
        for j in range(3):
            if state[i][j] == 0:
                return i, j


# Move 0
def move(state, new_i, new_j):

    i, j = find_zero(state)

    new_state = [row[:] for row in state]

    new_state[i][j], new_state[new_i][new_j] = \
        new_state[new_i][new_j], new_state[i][j]

    return new_state


# Generate possible moves
def get_neighbors(state):

    i, j = find_zero(state)

    neighbors = []

    # UP
    if i > 0:
        neighbors.append(move(state, i - 1, j))

    # DOWN
    if i < 2:
        neighbors.append(move(state, i + 1, j))

    # LEFT
    if j > 0:
        neighbors.append(move(state, i, j - 1))

    # RIGHT
    if j < 2:
        neighbors.append(move(state, i, j + 1))

    return neighbors


# Depth Limited Search
def depth_limited_search(state, goal, limit, depth, visited):

    print("Depth:", depth)
    print_puzzle(state)

    # Goal found
    if state == goal:
        print("GOAL FOUND!")
        print("Depth:", depth)
        return True

    # Depth limit reached
    if depth == limit:
        return False

    state_tuple = make_tuple(state)

    visited.add(state_tuple)

    # Generate next states
    neighbors = get_neighbors(state)

    for next_state in neighbors:

        next_tuple = make_tuple(next_state)

        # Avoid repeated states
        if next_tuple not in visited:

            found = depth_limited_search(
                next_state,
                goal,
                limit,
                depth + 1,
                visited
            )

            # If goal found, immediately exit
            if found:
                return True

    visited.remove(state_tuple)

    return False


# Iterative Deepening Search
def ids(start, goal):

    limit = 0

    while True:

        print("==============================")
        print("DEPTH LIMIT:", limit)
        print("==============================")

        visited = set()

        found = depth_limited_search(
            start,
            goal,
            limit,
            0,
            visited
        )

        # Goal found → EXIT IDS
        if found:
            print("IDS COMPLETED")
            return

        # Increase depth limit
        limit = limit + 1


# Start IDS
ids(start, goal)
```
